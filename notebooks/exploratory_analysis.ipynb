# Exploratory Data Analysis: Official BPS East Java (Jawa Timur) Dataset (2020–2024)

This notebook conducts exploratory analysis on the official Badan Pusat Statistik (BPS) dataset covering all **38 Regencies and Cities** in East Java across **2020, 2021, 2022, 2023, and 2024** (190 total observation rows).

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load final merged dataset
data_path = Path('../outputs/dataset_jatim_2020_2024.csv')
df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape} (Expected 190 rows x 13 columns)")
df.head()
```

## 1. Summary Statistics & Data Types

```python
df.info()
```

```python
df.describe().T
```

## 2. Regency & Year Coverage Check

```python
print(f"Total Regencies/Cities: {df['regency_city'].nunique()} (Expected 38)")
print(f"Years Covered: {sorted(df['year'].unique())} (Expected [2020, 2021, 2022, 2023, 2024])")
print(f"Grid Completeness Check: {len(df) == 38 * 5}")
```

## 3. Human Development Index (HDI / IPM) Trends by Region

```python
hdi_pivot = df.pivot(index='regency_city', columns='year', values='hdi')
hdi_pivot.head(10)
```

## 4. Open Unemployment Rate (TPT %) Trends

```python
tpt_pivot = df.pivot(index='regency_city', columns='year', values='unemployment_rate')
tpt_pivot.head(10)
```
